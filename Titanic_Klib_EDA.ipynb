# Titanic Dataset — Complete EDA Using Klib

This notebook uses the uploaded Kaggle Titanic `train.csv` dataset.

**Goal:** demonstrate a complete EDA workflow with Klib, including checking missing values, checking duplicates, cleaning the data with Klib, and generating automated EDA plots.

**Important:** Pandas is used only to load/check the dataset. EDA plots and automated cleaning are performed with Klib.

## 1. Install Klib

```python
!pip install -U klib
```

## 2. Import libraries

```python
import pandas as pd
import klib

pd.set_option('display.max_columns', None)
```

## 3. Load the uploaded Kaggle dataset

The file used here is the uploaded `train.csv`.

```python
df = pd.read_csv('train.csv')
df.head()
```

## 4. Basic dataset information

```python
print('Rows:', df.shape[0])
print('Columns:', df.shape[1])
print('\nColumns:')
print(df.columns.tolist())
```

## 5. Check data types

```python
df.dtypes
```

## 6. Missing-value check

First we identify missing values before cleaning.

```python
df.isnull().sum()
```

## 7. Klib missing-value visualization

Klib visualizes the missing-value pattern in the raw dataset.

```python
klib.missingval_plot(df)
```

## 8. Duplicate check

We check the raw dataset for completely duplicated rows.

```python
duplicate_count = df.duplicated().sum()
print('Duplicate rows:', duplicate_count)
```

## 9. Klib automated cleaning

`data_cleaning()` applies Klib's automated dataframe-cleaning process.

```python
df_clean = klib.data_cleaning(df)
df_clean.head()
```

## 10. Check the cleaned dataset

```python
print('Rows after cleaning:', df_clean.shape[0])
print('Columns after cleaning:', df_clean.shape[1])
print('\nMissing values after Klib cleaning:')
print(df_clean.isnull().sum())
```

## 11. Klib categorical analysis

Klib automatically visualizes categorical variables.

```python
klib.cat_plot(df_clean, top=4, bottom=4)
```

## 12. Klib numerical distribution analysis

Klib generates automated distributions for numerical variables.

```python
klib.dist_plot(df_clean)
```

## 13. Klib correlation analysis

The Titanic `survived` target is already numeric (0/1), so it can be used directly as Klib's correlation target.

```python
klib.corr_plot(df_clean)
```

## 14. Klib correlation with survival

```python
klib.corr_plot(df_clean, target='survived')
```

## 15. Final cleaned dataset

The cleaned dataframe is ready for further analysis or machine-learning work.

```python
df_clean.head()
```
